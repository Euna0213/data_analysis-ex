# 지도에 좌표 표시하기
- 공식문서 : https://python-visualization.github.io/folium/latest/getting_started.html#Choropleth-maps

```python
#uv add folium
```

```python
# 인터랙티브 지도를 손쉽게 만들 수 잇는 라이브러리
import folium
```

```python
# 디폴트 위경도 지정(포커싱), 지도사이트 Map 객체 생성
folium_map = folium.Map(location=[37.5662952, 126.9779451], zoom_start=16)
#folium.map -> 인터랙티브 지도 객체 생성
#location[위도, 경도] -> 지도 초기 중심 좌표
#zoom_start -> (1: 전세계), (18:건물단위) 숫자가 크면 클 수록 줌을 많이 당긴 느낌
```

```python
folium.Marker(location=[37.5662952, 126.9779451]).add_to(folium_map)
# 지도 위의 특정 좌표에 핀 모양 마커를 추가
# add_to : 생성한 마커를 지도에 연결하는 것 -> 이걸 입력해야 지도에 반영 됨
```

```python
popup = folium.Popup('서울시청 위치', max_width=300)
# 마커를 클릭했을 때 나타나는 말풍선(팝업) 텍스트 설정
# folium.Popup(팝업에 표시될 텍스트 내용, 팝업 창의 최대 너비(픽셀단위))

folium.Marker(location=[37.5662952, 126.9779451], popup=popup).add_to(folium_map)
#folium.Marker()에 popup 파라미터로 전달하면 마커 클릭 시 해당 팝업이 표시
```
